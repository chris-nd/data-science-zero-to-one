## L'attribut `out=`

```python
# le domaine de définition
X = np.linspace(0, 2*np.pi)
X
```

```python
Y =np.exp(np.sin(np.cos(X)))
Y
```

```python
plt.plot(X, Y);
```

```python
Y1 = np.cos(X)
Y2 = np.sin(X)
Y3 = np.exp(X)
plt.plot(X, Y1);
```

```python
plt.plot(X, Y2);
```

```python
plt.plot(X, Y3);
```

Presque tous les opérateurs numpy proposent un paramètre optionnel `out=` qui permet de spécifier un tableau déjà alloué, dans lequel ranger le résultat.

```python
X1 = np.cos(X)
np.sin(X1, out=X1)
np.exp(X1, out=X1)
plt.plot(X, X1);
```

## `dtype` structuré

```python
my_dtype = [
    # prenom est un string de taille 12
    ('prenom', '|S12'),
    # nom est un string de taille 15
    ('nom', '|S15'),
    # age est un entier
    ('age', int)
]

# un tableau qui contient des cellules de ce type
classe = np.array(
    # le contenu
    [ ( 'Jean', 'Dupont', 32),
      ( 'Daniel', 'Durand', 18),
      ( 'Joseph', 'Delapierre', 54),
      ( 'Paul', 'Girard', 20)],
    # le type
    dtype = my_dtype)

print(classe)
```

## Le stacking

Permet de créer un tableau plus grand avec la **juxtaposition** `hstack()` et la **superposition** `vstack()` pour les tableaux à 2 dimensions

```python
a = np.arange(1, 7).reshape(2, 3)
print(a)
```

```python
b = 10 * np.arange(1, 7).reshape(2, 3)
print(b)
```

```python
print(np.hstack((a, b)))
```

```python
print(np.vstack((a, b)))
```

## `np.concatenate()`

Pour agrandir les tableaux supérieurs à 2 dimensions

```python
a = np.ones((2, 3, 4))
print(a)
```

```python
b = np.zeros((2, 3, 2))
print(b)
```

```python
print(np.concatenate((a, b), axis=2))
```

## `np.tile()`

Permet de répéter un tableau dans toutes les directions

```python
motif = np.array([[0, 1], [2, 10]])
print(motif)
```

```python
print(np.tile(motif, 3))
```

```python
print(np.tile(motif, (2, 3)))
```

## `np.hsplit()` et `np.vsplit()`

Permet de découper un tableau en parties plus ou moins égales

```python
complet = np.arange(24).reshape(4, 6)
print(complet)
```

```python
h1, h2 = np.hsplit(complet, 2)
```

```python
print(h1)
```

```python
print(h2)
```

```python
complet = np.arange(24).reshape(4, 6)
print(complet)
```

```python
v1, v2 = np.vsplit(complet, 2)
```

```python
print(v1)
```

```python
print(v2)
```
