```python
import numpy as np
```

```python
L = list(range(1000))
```

## 1 - Bibliothèque Numpy

C'est la librairie de référence pour gérer et manipuler des tableaux multi-dimensionnels en python

```python
a = np.array(L)
```

```python
%timeit [x**2 for x in L]
```

```python
%timeit [x**2 for x in a]
```

```python
%timeit a**2
```

## 2 - Bibliothèque Pandas

C'est la librairie de référence qui permet d'ajouter des alias à des tableaux Numpy, ce qui sert à donner des labels aux lignes et aux colonnes des tableaux, afin de pouvoir accéder de manière plus intuitive et expréssive aux éléments de ce tableau

```python
# Seaborn (sn) vient avec quelques dataset (jeux de données) sur lesquels on peut jouer
import seaborn as sns

tips = sns.load_dataset('tips')
type(tips)
```

```python
tips.head()
```

```python
tips.groupby('sex', observed=True).mean(numeric_only=True)
```

```python
tips.groupby('day').mean(numeric_only=True)
```

```python
tips.groupby('time').mean(numeric_only=True)
```
