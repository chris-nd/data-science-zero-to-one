```python
import numpy as np
import pandas as pd
```

## Numpy

Numpy fournit deux types d'objets pour manipuler les dates: 

- `datetime64` pour représenter des dates et des heures

- `timedelta64` pour représenter des intervalles de temps (durées)

Ces deux objets sont codés en 64 bits

### `datetime64`

```python
# Avec une date au format ISO-8601
np.datetime64('2018-06-30')
```

```python
# Avec une date et une heure au format ISO-8601
np.datetime64('2018-06-30 08:35:23')
```

```python
# Avec une date et une heure au format ISO-8601.
# Et une précision à la nanoseconde (granularité)
np.datetime64('2018-06-30 08:35:23', 'ns')
```

### `timedelta64`

```python
# Les opérations avec des objets datetime64 renvoie un objet timedelta64
np.datetime64('2018-06-30 08:35:23') - np.datetime64('2018-06-20 08:37:23')
```

## Pandas

Pandas fournit trois types d'objets pour manipuler les dates:

- `Timestamp` pour représenter des dates et des heures

- `Timedelta` pour représenter des intervalles entre deux dates (durées)

- `Period` pour représenter une date associée à une durée (séries temporelles)

Ces objets `pandas` permettent un parsing beaucoup plus flexible qui ne se limite pas aux formats de date standard et on peut créer à partir de ces objets des index.

```python
pd.to_datetime('10 june 1973 8h30')
```

```python
pd.to_datetime(['10 june 1973 8h30', '22-JUNE-1973'])
```

```python
# Avec une fréquence d'un jour
index = pd.date_range('1 jan 2018', periods=1000, freq='D')
index
```

- Spécifier une fréquence

```python
# Avec une fréquence de 43 heures et 36 minutes
index = pd.date_range('1 jan 2018', periods=1000, freq='43h36min')
index
```

```python
# Avec une fréquence de 43 heures et 36 minutes
index = pd.date_range('1 jan 2018', periods=1000, freq='43h36min')
index
```

### Séries temporelles en pandas

> `NAT` : Not a Time

```python
date = '100/06/2018'

try:
    pd.to_datetime(date)
except ValueError as e:
    print(e)
```

```python
date = '100/06/2018'

try:
    pd.to_datetime(date, errors='raise')
except ValueError as e:
    print(e)
```

- gestion des erreur de parsing des dates

```python
pd.to_datetime(date, errors='coerce')
```

```python
d = pd.to_datetime(['jun 2018', '10/12/1980',
                    '25 january 2000', '100 june 1900'], errors='coerce')

print(d)
```

```python
d.fillna(pd.to_datetime('10 june 1980'))
```
