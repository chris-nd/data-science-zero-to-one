## Appliquer une opération à chaque élément d'un tableau numpy

Effectuer des calculs sur des tableaux numpy prennent moins de temps que sur des objets python natif comme les listes

```python
a = np.arange(1000)
```

```python
%timeit [x**2 + 2*x -1 for x in a]
```

```python
%timeit a**2 + 2*a -1
```

```python
a = np.arange(1, 1_000_000, dtype=np.float64)
```

```python
%timeit -r 1 -n 1 np.sqrt(a)
```

```python
%timeit -r 1 -n 1 np.sqrt(a, out=a)
```

## `np.mean()`

```python
a = np.arange(1, 10, dtype=np.float64).reshape(3, 3)
a
```

```python
a[1, 1] = np.nan
```

```python
a
```

Toutes opérations éffectués sur des NaN produisent en sortie un NaN

```python
np.mean(a)
```

## Les fonctions préfixées de `nan` ignorent les objets NaN

### `np.nanmean`

```python
np.nanmean(a)
```

### `np.nanmax`

```python
np.nanmax(a)
```
