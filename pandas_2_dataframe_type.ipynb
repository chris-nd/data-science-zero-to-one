```python
import numpy as np
import pandas as pd
```

```python
prenoms = ['liz', 'bob', 'bill', 'eve']
```

```python
age = pd.Series([25, 30, 35, 40], index=prenoms)
```

```python
taille = pd.Series([160, 175, 170, 180], index=prenoms)
```

```python
sexe = pd.Series(list("fhhf"), index=prenoms)
```

```python
df = pd.DataFrame(
    {
        'age': age,
        'taille': taille,
        'sexe': sexe
    },
)
```

```python
df1 = pd.DataFrame(np.ones((2, 2), dtype=np.int8),
                   index=list('ab'),
                   columns=list('xy')
                   )
```

```python
df1
```

```python
df2 = pd.DataFrame(np.ones((2, 2), dtype=np.int8),
                   index=list('ac'),
                   columns=list('xz')
                   )
```

```python
df2
```

### Résolution d'alignement d'index

```python
df1 + df2
```

- `fill_value` ne fonctionne que s'il manque une valeur à gauche ou à droite (pas si les deux sont absents)

```python
df1.add(df2, fill_value=0)
```

```python
df = df1.add(df2, fill_value=0)
df
```

- Remplacer les valeur `NaN` par une autre valeur

```python
df.fillna(-1)
```

- Retirer les lignes contenant des valeurs `NaN`

```python
df.dropna()
```
