```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
plt.ion();
```

```python
def background(n):
    i = np.arange(n)
    j = i.reshape((n,1))
    return i + 10 * j
```

## Indexation par une liste

```python
b = background(6)
print(b)
```

```python
print(b[[1, 3, 4]])
```

```python
print(b[:, [1, 3, 4]])
```

```python
b[:, [1, 3, 4]] = np.arange(1000, 1006).reshape(6, 1)
print(b)
```

### Application au codage de couleur dans une image

```python
N = 32
colors = 6

image = np.empty((N, N), dtype=np.int32)

for i in range(N):
    for j in range(N):
        image[i, j] = (i+j) % colors
```

```python
print(image)
```

```python
plt.imshow(image, cmap='gray')
```

```python
palette = np.array([
  [255, 255, 255], # 0 -> blanc
  [255, 0, 0],     # 1 -> rouge
  [0, 255, 0],     # 2 -> vert
  [0, 0, 255],     # 3 -> bleu
  [0, 255, 255],   # 4 -> cyan
  [255, 255, 0],   # 5 -> magenta
 ], dtype=np.uint8)
```

```python
print(palette)
```

```python
palette.shape
```

```python
print(image)
```

```python
image.shape
```

```python
print(palette[image])
```

```python
palette[image].shape
```

```python
plt.imshow(palette[image])
```

## Indexation multiple (par tuple)

```python
ix, iy = np.indices((4, 3))
```

```python
ix
```

```python
iy
```

```python
A = 10 * ix + iy
print(A)
```

```python
index1 = [[3, 1], [0, 1]]
index2 = [[2, 0], [0, 2]]
```

Lorsque *index1* et *index2* ont la même forme, on peut écrire

```python
print(A[index1, index2])
```

### Application à la recherche de maxima

```python
times = np.linspace(1000, 5000, num=5, dtype=int)
print(times)
```

```python
series = np.array([
    [10, 25, 32, 23, 12],
    [12, 8, 4, 10, 7],
    [100, 80, 90, 110, 120]
])
print(series)
```

```python
series.shape
```

```python
max_indices = np.argmax(series, axis=1)
print(max_indices)
```

```python
# series[[0, 1, 2], [2, 0 4]]
maxima = series[ range(series.shape[0]), max_indices ]
print(maxima)
```

```python
times[max_indices]
```
