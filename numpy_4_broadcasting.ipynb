## Manipulation de tableau avec Index

```python
def background(n):
    i = np.arange(n)
    j = i.reshape((n, 1))
    return i + 10 * j
```

```python
a5 = background(5)
print(a5)
```

```python
a5[1]
```

```python
print(a5[1][2])
```

Index de tuple


```python
print(a5[1, 2])
```

```python
a5[2][1] = 221
a5[3, 2] += 300
```

```python
print(a5)
```

```python
a5[1] = np.arange(100, 105)
print(a5)
```

```python
a5[4] = 400
print(a5)
```

```python
a8[1::3]
```

Le slicing peut servir à extraire des blocs

## Masques

```python
width = 128
center = width / 2

ix, iy = np.indices((width, width))
image = (ix - center)**2 + (iy - center)**2
plt.imshow(image, cmap='gray');
```

```python
print(ix)
```

```python
print(iy)
```

```python
print(image)
```

```python
ix + iy
```

```python
(ix + iy) % 8
```

```python
(ix + iy) % 8 <= 5
```

```python
rayures = (ix + iy) % 8 <= 5
plt.imshow(rayures, cmap='gray')
```

### L'opérateur `not` doit être remplacé par `~` ou `np.logical_not()`

```python
try:
    anti_rayures = not rayures
except Exception as e:
    print(f'OOPS {type(e)} - {e}')
```

```python
anti_rayures = np.logical_not(rayures)
plt.imshow(anti_rayures, cmap='gray');
```

```python
print(anti_rayures)
```

```python
anti_rayures = ~rayures
plt.imshow(anti_rayures, cmap='gray')
```

```python
print(anti_rayures)
```

```python
image*rayures
```

```python
plt.imshow(image*rayures, cmap='gray')
```

```python
image*anti_rayures
```

```python
plt.imshow(image*anti_rayures, cmap='gray');
```

### `np.close`

```python
big_circle = np.isclose(image, 64**2, 10/100)
big_circle
```

```python
plt.imshow(big_circle, cmap='gray');
```

```python
small_circle = np.isclose(image, 32**2, 10/100)
plt.imshow(small_circle, cmap='gray')
```
