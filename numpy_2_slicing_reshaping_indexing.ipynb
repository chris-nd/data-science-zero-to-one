## Le redimensionnement avec `reshape`

```python
a = np.random.randint(1, 10, size=(4, 4))
a
```

```python
b = a.reshape(8, 2)
b
```

```python
a
```

```python
# Le redimensionnement avec reshape nécessite de conserver une compatibilité entre le nombre de lignes et colonnes (nombre de cellules).
try:
    a.reshape((5,4))
except Exception as e:
    print(f'OOPS {type(e)} {e}')
```

### Dimensions supérieures avec `reshape()`

```python
shape = (2, 3, 4)
size = reduce(lambda x, y: x*y, shape)

data = np.arange(size)
data
```

```python
d3 = data.reshape(shape)
d3
```

```python
shape = (2, 3, 4, 5)
size = reduce(lambda x, y: x*y, shape)
size
```

```python
data = np.arange(size)

d4 = data.reshape(shape)
d4
```

### `meshgrid`

```python
Xticks, Yticks = (np.linspace(-np.pi, np.pi, num=10),
                  np.linspace(3*np.pi, 5*np.pi, num=10))
```

```python
X, Y = np.meshgrid(Xticks, Yticks)

X.shape, Y.shape
```

```python
X
```

```python
Y
```

```python
Z = np.cos(X) + np.cos(Y)**2
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_wireframe(X, Y, Z);
```
