## Jointure des DataFrames avec `pd.merge()`

Par défaut, `merge` fait un **inner join**

```python
df1 = pd.DataFrame({'personnel': ['Bob', 'Lisa', 'Sue'],
                    'groupe': ['SAF', 'R&D', 'RH']})

df1
```

```python
df2 = pd.DataFrame({'personnel': ['Lisa', 'Bob', 'Sue'],
                    'date embauche': ['2004', '2008', '2014']})

df2
```

```python
pd.merge(df1, df2)
```


Il existe trois type de merges :

- one-to-one, c’est le merge lorqu’il n’y a pas de labels dupliqués dans les colonnes utilisées comme critère de merge ;

- many-to-one, c’est le merge lorsque l’une des deux colonnes contient des labels dupliqués, dans ce cas, on applique la stratégie one-to-one pour chaque label dupliqué, donc les entrées dupliquées sont préservées ;

- many-to-many, c’est la stratégie lorsqu’il y a des entrées dupliquées dans les deux colonnes. Dans ce cas, on fait un produit cartésien des lignes.

```python
df1 = pd.DataFrame({'patient': ['Bob', 'Lisa', 'Sue'],
                    'repas': ['SS', 'SS', 'SSR']})
df1
```

```python
df2 = pd.DataFrame({'repas': ['SS', 'SSR'],
                    'explication': ['sans sel', 'sans sucre']})
df2
```

```python
pd.merge(df1, df2)
```

```python
df1 = pd.DataFrame({'patient': ['Bob', 'Lisa', 'Sue'],
                    'repas': ['SS', 'SS', 'SSR']})

df1
```

```python
df2 = pd.DataFrame({'repas': ['SS', 'SS', 'SSR'],
                    'explication': ['sans sel', 'légumes', 'sans sucre']})

df2
```

```python
pd.merge(df1, df2)
```

```python
df1 = pd.DataFrame({'employee': ['Bob', 'Lisa', 'Sue'],
                    'group': ['Accounting', 'Engineering', 'HR']})

df1
```

```python
df2 = pd.DataFrame({'employee': ['Lisa', 'Bob', 'Sue'],
                    'hire_date': [2004, 2008, 2014]})

df2
```

```python
pd.merge(df1, df2, on='employee')
```

```python
df1 = pd.DataFrame({'employee': ['Bob', 'Lisa', 'Sue'],
                    'group': ['Accounting', 'Engineering', 'HR']})
df1
```

```python
df2 = pd.DataFrame({'name': ['Lisa', 'Bob', 'Sue'],
                    'hire_date': [2004, 2008, 2014]})
df2
```

```python
m = pd.merge(df1, df2, left_on='employee', right_on='name')
m
```

### Supprimer une colonne

```python
m.drop('name', axis=1)
```

### Merge externe, à gauche et à droite

```python
df1 = pd.DataFrame({'name': ['Bob', 'Lisa', 'Sue'],
                    'pulse': [70, 63, 81]})
df1
```

```python
df2 = pd.DataFrame({'name': ['Eric', 'Bob', 'Marc'],
                    'weight': [60, 100, 70]})
df2
```

```python
pd.merge(df1, df2)
```

```python
pd.merge(df1, df2, how='outer')
```

```python
pd.merge(df1, df2, how='left')
```

```python
pd.merge(df1, df2, how='right')
```

## Regrouper des DataFrames avec `pd.groupby()`

```python
import seaborn as sns
```

```python
titanic = sns.load_dataset('titanic').loc[:, ['survived', 'sex', 'class']]

titanic.head(10)
```

```python
titanic.shape
```

```python
titanic.loc[:, 'survived'].mean()
```

```python
titanic.loc[titanic.loc[:, 'class']=='First','survived'].mean()
```

```python
titanic.groupby('class')
```

```python
g = titanic.groupby(['class', 'sex']).mean()

g
```

```python
g.index
```

## Faire pivoter un DataFrame avec ``

```python
titanic.pivot_table('survived',
                    aggfunc=np.mean,
                    index='class',
                    columns='sex')
```
