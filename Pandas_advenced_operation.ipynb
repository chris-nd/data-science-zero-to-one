## Jointure des DataFrames avec `pd.merge()`

Par défaut, `merge` fait un **inner join**

```python
df1 = pd.DataFrame({'personnel': ['Bob', 'Lisa', 'Sue'],
                    'groupe': ['SAF', 'R&D', 'RH']})

df1
```

```python
df2 = pd.DataFrame({'personnel': ['Lisa', 'Bob', 'Sue'],
                    'date embauche': ['2004', '2008', '2014']})

df2
```

```python
pd.merge(df1, df2)
```


Il existe trois type de merges :

- one-to-one, c’est le merge lorqu’il n’y a pas de labels dupliqués dans les colonnes utilisées comme critère de merge ;

- many-to-one, c’est le merge lorsque l’une des deux colonnes contient des labels dupliqués, dans ce cas, on applique la stratégie one-to-one pour chaque label dupliqué, donc les entrées dupliquées sont préservées ;

- many-to-many, c’est la stratégie lorsqu’il y a des entrées dupliquées dans les deux colonnes. Dans ce cas, on fait un produit cartésien des lignes.

```python
df1 = pd.DataFrame({'patient': ['Bob', 'Lisa', 'Sue'],
                    'repas': ['SS', 'SS', 'SSR']})
df1
```

```python
df2 = pd.DataFrame({'repas': ['SS', 'SSR'],
                    'explication': ['sans sel', 'sans sucre']})
df2
```

```python
pd.merge(df1, df2)
```

## Regrouper des DataFrames avec `pd.groupby()`

```python
import seaborn as sns
```

```python
d = pd.DataFrame({'key': list('ABCABC'), 'val': range(6)})
d
```

`groupby` produit un nouvel objet de groupe, qui permet d'appliquer des opérations sur chaque groupe de lignes.

```python
g = d.groupby('key')
g
```

```python
g.sum()
```

```python
g = d.groupby('key', as_index=False)
g.sum()
```

```python
d = pd.DataFrame({'key': list('ABCABC'),
                  'val1': range(6),
                  'val2' : range(100, 106)})
d
```

```python
g = d.groupby('key')
```

g.groups donne accès au dictionnaire des groupes, les clefs sont le nom du groupe et les valeurs les index des lignes
appartenant au groupe.

```python
g.groups
```

```python
g.get_group('A')
```

```python
g.sum()['val2']
```

```python
g['val2'].sum()
```

On peut également itérer sur les groupes avec un boucle for classique

```python
tips = sns.load_dataset("tips")
```

```python
tips.head()
```

```python
g = tips.groupby('day')
g.groups
```

```python

for (group, index) in g:
    print(f"On {group} the mean tip is {index['tip'].mean():.3}")
```

on groupe par jour et on extrait uniquement la colonne 'total_bill' pour chaque groupe

```python
g = tips.groupby('day')['total_bill']
```

## Fonction dispatch

on demande à pandas d'afficher les float avec seulement deux chiffres après la virgule

```python
pd.set_option('display.float_format', '{:.2f}'.format)
```

```python
g.describe()
```

Il y a tout de même un grand nombre de méthodes définies directement sur l'objet produit par le groupby

```python
methods = [x for x in dir(g) if not x.startswith('_')]
f"Le type {type(g).__name__} expose {len(methods)} méthodes."
```

```python
columns = 7
nb_methods = len(methods)
nb_pad = (columns - nb_methods % columns) % columns

array = np.array(methods + nb_pad * ['']).reshape((columns, -1))
```

```python
pd.DataFrame(data=array.transpose())
```

La méthode `aggregate` (dont l’alias est `agg`) est la méthode, qui permet d’appliquer une fonction (ou liste de fonctions) à chaque groupe avec la possibilité d’appliquer une fonction à une colonne spécifique du groupe.

```python
numeric_columns = tips.describe().columns

tips.groupby('day')[numeric_columns].agg(['mean', 'std'])
```

```python
tips.groupby('day')[numeric_columns].agg([np.mean, np.std])
```

```python
tips.groupby('day').agg({'tip': np.mean, 'total_bill': np.std})
```

```python
d = pd.DataFrame({'key': list('ABCABC'), 
                  'val1': range(6),
                  'val2' : range(100, 106)})
d
```

```python
d.groupby('key').sum()
```

```python
d.groupby('key').filter(lambda x: x['val1'].sum() > 3)
```
