```python
import numpy as np
import pandas as pd
from pandas.conftest import axis
```

```python
df1 = pd.DataFrame(np.random.randint(1, 10, size=(2, 2)),
    index=list('xy'),
    columns=list('ab')
)

df1
```

```python
df2 = pd.DataFrame(np.random.randint(1, 10, size=(2, 2)),
    index=list('zt'),
    columns=list('ab')
)

df2
```

## Concaténer des DataFrames avec `pd.concat()`

```python
pd.concat([df1, df2])
```

### Changer le sens de la concaténation pour une orientation horizontale

```python
df1 = pd.DataFrame(np.random.randint(1, 10, size=(2, 2)),
    index=list('xy'),
    columns=list('ab')
)

df1
```

```python
df2 = pd.DataFrame(np.random.randint(1, 10, size=(2, 2)),
    index=list('xy'),
    columns=list('cd')
)

df2
```

```python
pd.concat([df1, df2], axis=1)
```

## Jointure des DataFrames avec `pd.merge()`

```python
df1 = pd.DataFrame({'personnel': ['Bob', 'Lisa', 'Sue'],
                    'groupe': ['SAF', 'R&D', 'RH']})

df1
```

```python
df2 = pd.DataFrame({'personnel': ['Lisa', 'Bob', 'Sue'],
                    'date embauche': ['2004', '2008', '2014']})

df2
```

```python
pd.merge(df1, df2)
```

## Regrouper des DataFrames avec `pd.groupby()`

```python
import seaborn as sns
```

```python
titanic = sns.load_dataset('titanic').loc[:, ['survived', 'sex', 'class']]

titanic.head(10)
```

```python
titanic.shape
```

```python
titanic.loc[:, 'survived'].mean()
```

```python
titanic.loc[titanic.loc[:, 'class']=='First','survived'].mean()
```

```python
titanic.groupby('class')
```

```python
g = titanic.groupby(['class', 'sex']).mean()

g
```

```python
g.index
```

## Faire pivoter un DataFrame avec ``

```python
titanic.pivot_table('survived',
                    aggfunc=np.mean,
                    index='class',
                    columns='sex')
```
