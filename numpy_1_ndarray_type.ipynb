### Universal function ou ufunc

Ces fonctions implémentent un mécanisme qui permet à la fonction de s'exécuter sur chaque terme d'un tableau par exemple.

#### Passer d'une fonction scalaire à une ufunc avec le décorateur `@np.vectorize`

```python
@np.vectorize
def scalar_function(x):
    return x**2 + 2*x + (1 if x <= 0 else 10)
```

```python
X = np.linspace(-5, 5, 1000)
Y = scalar_function(X)
plt.plot(X, Y)
```

## La différence de taille mémoire entre un tableau dynamique python(list) et un tableau numpy

```python
from sys import getsizeof
array_py = [0 for _ in range(10**6)]
getsizeof(array_py)
```

```python
array_np = np.zeros(10**6, dtype=np.int8)
getsizeof(array_np)
```
