## Création d'une Series avec `np.Series()`

Une `Series` a une certaine similarité avec un dictionnaire python

```python
S = pd.Series([x**2 for x in range(10)])
S
```

```python
S = pd.Series([x**2 for x in range(10)], dtype='int8')
S
```

```python
S = pd.Series(data=[x**2 for x in range(10)],
              index=[x for x in 'abcdefghij'],
              dtype='int8')
S
```

- Avec un dictionnaire

```python
d = {k: v**2 for k, v in zip('abcdefghij', range(10))}
print(d)
```

```python
S = pd.Series(d, dtype='int8')
S
```

```python
print(S['f'])
```

```python
print(S.loc['f'])
```

L’index d’une Series est un objet implémenté sous la forme d’un ndarray de numpy, mais qui ne peut contenir que des objets hashables (pour garantir la performance de l’accès).

```python
print(S.index)
```

Les valeurs d'une Series sont accessibles avec l'attribut `values`qui renvoie un objet ndarray

```python
print(S.values)
```

Mais une Series a également une interface de dictionnaire à laquelle on accède de la manière suivante:

```python
S.keys()
```

```python
for k, v in S.items():
    print(k, v)
```

Le test d'appartenance est possible sur les index

```python
'a' in S, 'z' in S
```

Le slicing sur les labels ne fonctionne pas s'ils sont dubliqués et ne sont pas triés

```python
animaux = ['chien', 'chat', 'chat', 'chien', 'poisson']
proprio = ['eve', 'bob', 'eve', 'bill', 'liz']
```

```python
s = pd.Series(animaux, index=proprio)
s
```

```python
# Ceci lancera une erreur
s.loc['eve': 'liz']
```

```python
s = s.sort_index()
s
```

```python
s.loc['eve': 'liz']
```
