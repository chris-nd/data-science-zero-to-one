## Manipulation des matrices et vecteurs

### Produit Matriciel avec `np.dot()`, `dot()` ou `@` (opérateur *ad hoc* pour numpy)

```python
m1 = np.array([[1, 1],
               [2, 2]])
print(m1)
```

```python
m2 = np.array([[10, 10],
               [30, 40]])
print(m2)
```

```python
np.dot(m1, m2)
```

```python
m1.dot(m2)
```

```python
m1 @ m2
```

```python
m2 @ m1
```

### Produit scalaire avec `np.dot()`, `dot()` ou `@`

```python
v1 = np.array([1, 2, 3])
print(v1)
```

```python
v2 = np.array([4, 5, 6])
print(v2)
```

```python
scalar_product = np.dot(v1, v2)
```

```python
scalar_product = v1.dot(v2)
print(scalar_product)
```

```python
scalar_product = v1 @ v2
print(scalar_product)
```

### Matrice diagonales avec `np.diag()`

```python
M = np.arange(4) + 10 * np.arange(4)[:, np.newaxis]
print(M)
```

```python
# Extraire la diagonale principale
D = np.diag(M)
print(D)
```

```python
# Construire une matrice à partir d'une diagonale
M2 = np.diag(D)
print(M2)
```

### Déterminant de matrice avec `np.linalg.det()`

```python
M = np.array([[0, -1],
              [1, 0]])
print(M)
```

```python
np.linalg.det(M) == 1
```

### Valeur et vecteur propre de matrice avec `np.linalg.eig`

- Un vecteur propre est une flèche (un vecteur) dont la direction ne change pas lorsqu'on lui applique une transformation géométrique ou une matrice.

 - La valeur propre est le nombre qui indique de combien ce vecteur est étiré, réduit ou inversé lors de cette transformation.

 Imaginez une transformation (comme un zoom ou une déformation) dans l'espace.
 - La plupart des vecteurs changent à la fois de taille et de direction.

  - Les vecteurs propres ont un comportement spécial : ils gardent exactement la même direction, ils sont juste multipliés par un facteur d'échelle.

  - Ce facteur d'échelle est la valeur propre (souvent notée `lambda`).

Obtenir des valeurs et vecteurs propres

```python
S = np.array([[0, 1],
              [1, 0]])
```

```python
values, vectors = np.linalg.eig(S)
```

```python
values
```

```python
vectors
```

### Systèmes d'équations avec `np.linalg.solve`

```python
x, y, z = 1, 2, 3
```

```python
3*x + 2*y + z
```

```python
2*x + 3*y + 4*z
```

```python
5*x + 2*y + 5*z
```

```python
coefficients= np.array([
    [3, 2, 1],
    [2, 3, 4],
    [5, 2, 6],
])
```

```python
constants = [
    10,
    20,
    27,
]
```

```python
X, Y, Z = np.linalg.solve(coefficients, constants)
```

```python
print(X)
```

```python
print(Y)
```

```python
print(Z)
```
